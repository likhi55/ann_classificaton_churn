```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import pandas as pd
import pickle
```

```python
# Load the trained model
model =load_model("model.h5")

# Load the scaler
with open('label_encoder_gender.pkl', 'rb') as f:
    label_encoder = pickle.load(f)

with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

with open('onehot_encoder_geography.pkl', 'rb') as f:
    onehot_encoder = pickle.load(f)
```

```python
input_data = {
    "CreditScore": 600,
    "Geography": "France",
    "Gender": "Male",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000
}
```

```python
# One-hot encode the 'Geography' feature
geography_encoded = onehot_encoder.transform([[input_data['Geography']]]).toarray()
geography_df = pd.DataFrame(geography_encoded, columns=onehot_encoder.get_feature_names_out(['Geography']))

input_df = pd.DataFrame([input_data])
```

```python
#Encode categorical features
input_df['Gender'] = label_encoder.transform(input_df['Gender'])
```

```python
#Concatenate the one-hot encoded geography features with the rest of the input data
input_df = pd.concat([input_df.drop('Geography', axis=1), geography_df], axis=1)
```

```python
# Scaling the input data
input_scaled = scaler.transform(input_df)
input_scaled
```

```python
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_probability = prediction[0][0]

if prediction_probability > 0.5:
    print('The customer is likely to churn.')
else:    
    print('The customer is not likely to churn.')
```
